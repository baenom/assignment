```python
import random

def f(x):
    return (x-3)**2+4

def dx_f(x):
    return 2*(x-3)

L = 0.001

iterations = 100_000

x = random.randint(-15,15)

print(x)

for i in range(iterations):
    d_x = dx_f(x)
    x -= L*d_x

print(x,f(x))
```
